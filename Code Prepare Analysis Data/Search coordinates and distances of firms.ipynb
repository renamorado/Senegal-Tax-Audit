```python
from selenium import webdriver
from time import sleep

import pandas as pd
```

```python
# Open page 
driver = webdriver.Chrome()
driver.get("https://www.google.com/maps/@14.6705245,-17.4357241,17.83z")
sleep(2)

# set a search departing from DGID in Dakar Plateau
def searchplace(x, y):
    Place = driver.find_element('xpath','//*[@id="searchboxinput"]')    
    Place.send_keys(x)
    Place.send_keys("\n")
    sleep(3)
    Place2 = driver.find_element('xpath', '//*[@id="sb_ifc50"]/input')
    Place2.clear()
    Place2.send_keys(y)   
    Place2.send_keys("\n")   

searchplace("KM 22 RTE DE RUFISQUE", "31 Rue de Thiong, Dakar, Senegal")

sleep(5)
#Click on settings to choose date and time
hourtime = driver.find_element('xpath', '//*[@id="QA0Szd"]/div/div/div[1]/div[2]/div/div[1]/div/div/div[2]/span/div/div/div/div[2]')
hourtime.click()

#Click on settings to "depart at"
hourtime = driver.find_element('xpath', '//*[@id=":3"]/div')
hourtime.click()
sleep(1)

#Choose 3 PM
hour = driver.find_element('xpath', '//*[@id="QA0Szd"]/div/div/div[1]/div[2]/div/div[1]/div/div/div[2]/div/span[1]/input')
hour.clear()
hour.send_keys("3:00 PM")   
sleep(1)

#Choose Monday September 6th, 2023
date = driver.find_element('xpath', '//*[@id="QA0Szd"]/div/div/div[1]/div[2]/div/div[1]/div/div/div[2]/div/span[2]/span[2]')
date.click()
date = driver.find_element('xpath', '//*[@id=":2b"]')
date.click()
```
